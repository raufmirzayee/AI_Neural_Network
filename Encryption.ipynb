```python
# Import the Fernet encryption library
from cryptography.fernet import Fernet
key = Fernet.generate_key()
cipher = Fernet(key)
```

```python
name = input("Customer Name: ")
email = input("Email Address: ")
```

```python
encrypted_name = cipher.encrypt(name.encode())

print("Encrypted Name:")
print(encrypted_name)

```

```python
encrypted_email = cipher.encrypt(email.encode())

print("Encrypted Email:")
print(encrypted_email)

```

```python
decrypted_name = cipher.decrypt(encrypted_name)

print("Decrypted name:")
print(decrypted_name.decode())
```

```python
decrypted_email = cipher.decrypt(encrypted_email)

print("Decrypted Email:")
print(decrypted_email.decode())
```
